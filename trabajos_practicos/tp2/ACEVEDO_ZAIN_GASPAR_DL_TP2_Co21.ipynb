### Creación de datasets

En esta sección se realiza el preprocesamiento de los dos datasets de *train* y *validation*.

Tal como se menciona en el post de [DataQuest - Computer Vision in Pytorch](https://www.dataquest.io/blog/computer-vision-in-pytorch-part-2/#:~:text=This%20involves%20creating%20processing%20pipelines%20using%20torchvision.,and%20for%20validation/test%20data%20(which%20does%20not).), se utilizará un preprocesamiento para *train* y otro para *validate*, teniendo el primero operaciones de data augmentation (como girar verticalmente la imagen, o rotarla una determinada cantidad de grados). El motivo por el cual se utilizan dos preprocesamientos distintos es que para *train* se recomienda tener la mayor cantidad de *variabilidad* en las imágenes, a fin de que el modelo pueda generalizar, mientras que en *validate* se recomienda usar las imágenes originales, con un preprocesamiento mínimo (escala de grises, normalización).

Se realizarán operaciones similares a las implementadas en el paper de [Deep convolutional neural network architecture for facial emotion recognition](https://pmc.ncbi.nlm.nih.gov/articles/PMC11784769/#sec4):

- ***Escala de grises***: el paper recomienda el uso de *escala de grises*, ya que permite reducir la complejidad y dimensionalidad, reduce el *ruido* como así también incrementa la eficiencia computacional (se utiliza un solo canal).  
  - Se utiliza [torchvision.transforms.Grayscale](https://docs.pytorch.org/vision/0.9/transforms.html#torchvision.transforms.Grayscale).
- ***Resizing***: tal como recomienda el paper, se utilizará un *unico* **target size**. En este caso se utiliza un tamaño de imágenes de $(200\space x\space 200)$ tanto en *train* como *validate*, a fin de asegurar consistencia.
  - Se utiliza [torchivision.transforms.Resize](https://docs.pytorch.org/vision/0.9/transforms.html#torchvision.transforms.Resize).
- ***Escalado***: otra recomendación es la de escalar los píxeles, dividiendo su valor por $255$, obteniendo valores entre $0$ y $1$, lo cual permite acelerar la ***tasa de convergencia*** del modelo
  - Se utiliza [torchvision.transforms.ToTensor](https://docs.pytorch.org/vision/0.9/transforms.html#torchvision.transforms.ToTensor), ya que convierte la imagen en un ***tensor*** en el rango deseado $[0, 1]$.
- ***Data Augmentation***: se recomienda también, a modo de mejorar la generalización del modelo, utilizar técnicas de data augmentation, tales como aleatoriamente girar vertical u horizontalmente las imágenes.
  - Para esto, se utilizará [torchivision.transforms.RandomHorizontalFlip](https://docs.pytorch.org/vision/main/generated/torchvision.transforms.RandomHorizontalFlip.html), con una probabilidad de $0.3$ de voltear la imágen horizontalmente.
  - También se utilizará [torchivision.transforms.RandomVerticalFlip](https://docs.pytorch.org/vision/0.15/generated/torchvision.transforms.RandomVerticalFlip.html#torchvision.transforms.RandomVerticalFlip), con una probabilidad de $0.3$ de voltear la imágen verticalmente.
- ***Normalización***: se opta por implementar normalización, ya que esta permite mejorar la convergencia del modelo, como así también previene el *overfitting*. Esto se logra escalando las features a un *rango similar*, al utilizar el *desvío* y la *media*.
  - En este caso, utilizamos *media* y *desvío* de $0.5$, lo cual nos deja en el rango $[-1, 1]$.
  - [Fuente - post en Medium](https://medium.com/@piyushkashyap045/image-normalization-in-pytorch-from-tensor-conversion-to-scaling-3951b6337bc8).
  - Se utiliza [torchvision.transforms.Normalize](https://docs.pytorch.org/vision/0.9/transforms.html#torchvision.transforms.Normalize).

Se estudian también otras transformaciones recomendadas por PyTorch, tales como:

- [transforms.RandomRotation](https://docs.pytorch.org/vision/main/generated/torchvision.transforms.RandomRotation.html): permite rotar la imagen un cantidad determinada de grados.
- [transforms.ColorJitter](https://docs.pytorch.org/vision/main/generated/torchvision.transforms.ColorJitter.html): permite cambiar aleatoriamente el brillo, contraste y saturación de la imagen.

```python
def calcular_class_weights(dataset):
  class_counts = np.array([0] * len(dataset.classes))
  for _, label in dataset.samples:
      class_counts[label] += 1

  total_samples = len(dataset)
  class_weights = total_samples / class_counts

  weights_tensor = torch.tensor(class_weights, dtype=torch.float)
  return weights_tensor
```

```python
IMG_SIZE=200
```

```python
transformations_train = transforms.Compose([
    transforms.Grayscale(num_output_channels=1),
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(p=0.3),
    transforms.RandomVerticalFlip(p=0.3),
    transforms.RandomRotation(15),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])
```

```python
transformations_validate = transforms.Compose([
    transforms.Grayscale(num_output_channels=1),
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])
```

```python
train_dataset = datasets.ImageFolder(root=TRAIN_PATH, transform=transformations_train)
validation_dataset = datasets.ImageFolder(root=VALIDATION_PATH, transform=transformations_validate)
```

```python
weights_train = calcular_class_weights(train_dataset)

device = torch.device("cuda:0") if torch.cuda.is_available() else torch.device("cpu")
weights_train.to(device)
```

```python
weightedSampler = WeightedRandomSampler(weights=weights_train, num_samples=len(weights_train), replacement=True)
```

### Definición del modelo

El modelo a definir se caracterizará por utilizar bloques de convolución conformados por:

1. Una capa de *convolución*.
1. Una capa de *batch normalization*.
1. Una capa de *No linealidad*.
1. Una capa de *pooling*.

***Batch Normalization*** se utilizará con el fin de evitar overfitting de la red. Además, permite que haya menos dependencia a la inicialización de los pesos de la red. Para implementarlo, se utilizará la función [BatchNorm2d](https://docs.pytorch.org/docs/stable/generated/torch.nn.BatchNorm2d.html) de PyTorch.
  - Fuente: [Why batch normalization matters for deep learning - Towards data science](https://towardsdatascience.com/why-batch-normalization-matters-for-deep-learning-3e5f4d71f567/#:~:text=Advantages:,such%20as%20a%20dropout%20layer.).
  - Fuente: [What is batch normalization in deep learning - Milvus](https://milvus.io/ai-quick-reference/what-is-batch-normalization-in-deep-learning).

En la capa de ***No linealidad*** se utilizará la función de activación ***ReLU*** por los siguientes motivos:

- Las CNN que utilizan *ReLU* se entrenan varias veces más rápido que aquellas que utilizan otras funciones de activación, tales como *tanh*, como así también tienen muy buena performance.
  - Fuente: [ImageNet Classification with Deep Convolutional
Neural Networks](https://proceedings.neurips.cc/paper_files/paper/2012/file/c399862d3b9d6b76c8436e924a68c45b-Paper.pdf).
  - Fuente: [Facial Emotion Recognition using CNN in PyTorch](https://arxiv.org/html/2312.10818v1).
- Permite extraer patrones NO lineales en los datos de entrada al utilizarse en las CNN. Además, introducen poco bias y alta varianza.
  - Fuente: [Activation Functions in CNN - Medium](https://medium.com/@vishnuam/activation-functions-in-convolutional-neural-networks-cnn-b1114a1b9b4d).

En la capa de ***pooling*** se utilizará ***Max Pooling*** ya que al enfocarse en tomar el valor máximo en cada región, permite a la CNN ejecutarse de manera rápida, como así también enfocarse en las features más importantes de dicha región, lo cual a su vez evita que se enfoque en el ruido, es decir, evita el overfitting.
  - Fuente: [Max Pooling in CNNs](https://medium.com/@onally.sourour9/max-pooling-in-cnns-why-it-matters-and-how-it-works-60c590bca8b9).
  - Fuente: [Facial Emotion Recognition using CNN in PyTorch](https://arxiv.org/html/2312.10818v1).

A la hora de realizar el entrenamiento, se tendrá en cuenta lo siguiente:
- *Función de costo*: se utilizará [CrossEntropyLoss](https://docs.pytorch.org/docs/stable/generated/torch.nn.CrossEntropyLoss.html), ya que es un problema de clasificación de más de dos clases ($7$ en este caso).
- *Optimizador*: se utilizará ***Adam weighted*** ([AdamW](https://docs.pytorch.org/docs/stable/generated/torch.optim.AdamW.html) en PyTorch) ya que presenta los mismos beneficios que *Adam*, pero permite "desacoplar" el *weight decay*, permitiendo que actúe como una técnica de regularización pura sin afectar la dinámica de la tasa de aprendizaje.
  - *Learning Rate*: se estudiarán los siguientes valores mediante optimización de hiperparámetros $0.001$, $0.0005$, $0.0001$, $0.00005$, $0.00001$.
  - *Weight Decay*: se estudiarán los siguientes valores mediante optimización de hiperparámetros $0.01$, $0.005$, $0.001$, $0.0005$, $0.0001$.

En cuanto a la arquitectura de la CNN, se opta por estudiar un modelo con tres capas convolucionales, dos capas de dropout, y una capa fully-connected:

- ***Dropout***: habrá una capa luego de cada bloque de convolución. Éstas utilizarán un dropout de $0.5$, lo cual hace que las neuronas de la capa anterior "se apaguen" con una probabilidad de $50\%$, permitiendo reducir el ***overfitting*** en el training.
- ***Fully-connected***: Se ubica luego de la última capa de convolución. En ésta se aplanan las características extraídas en la capa anterior en un vector unidimensional, lo cual permite realizar la clasificación deseada.

***Capas/bloques de convolución***

Como ya se mencionó anteriormente, estas tendrán una convolución, un ***batch normalization***, una función de activación ***ReLU*** y un ***Max Pooling***.

- ***Convolución***
  - Se decide estudiar tres capas, en donde el output de cada una (cantidad de kernels) corresponde a los valores $16$, $32$ y $64$. Se eligió el valor $16$ aleatoriamente, mientras que los otros corresponden al duplicado de la capa anterior, ya que esto permite a nuestra red aumentar su capacidad de aprendizaje, más precisamente, duplicar al de la capa/bloque convolución anterior.
  - ***Stride***: Se opta por utilizar un valor de $1$, a fin de que el kernel explore todas las posiciones posibles (respecto al input analizado).
  - ***Padding***: Se opta por utilizar un `same` padding, a fin que la operación de convolución no afecte el tamaño de la imagen analizada (input y outputs iguales). Fuente: [Material complementario de la clase 5](https://github.com/FIUBA-Posgrado-Inteligencia-Artificial/aprendizaje_profundo/blob/master/clase_5/teoria/A%20guide%20to%20conv%20DUMOULIN%20VISIN.pdf).
  - ***Kernel size***: en cada capa de convolución se utilizarán kernels del mismo tamaÑo (n x n). El valor se determinará a partir del resultado de la `optimización de hiperparámetros`. El rango a explorar consta de valores impares (recomendado al utilizar padding `same` en [Material complementario de la clase 5](https://github.com/FIUBA-Posgrado-Inteligencia-Artificial/aprendizaje_profundo/blob/master/clase_5/teoria/A%20guide%20to%20conv%20DUMOULIN%20VISIN.pdf)), más precisamente, $3$, $5$, $7$, $11$, $13$ y $17$.
- ***Max Pooling***: se opta por utilizar un [MaxPool2d](https://docs.pytorch.org/docs/stable/generated/torch.nn.MaxPool2d.html) con un kernel size igual a $2$ y el resto de los valores por defecto. Considerando que los inputs corresponden a las matrices de las imágenes a analizar (cuadradas, o de nxn), la dimensión de salida se define como:
   - $W_{out} = \lfloor \frac{W_{in} + 2 * padding - dilation * (kernel\_size - 1) -1 }{stride}\rfloor + 1$
   - con $dilation = 1$, $padding = 0$ y $stride = kernel size$ (para esta capa se usa None)
   - $W_{out} = \lfloor \frac{W_{in} + 2 * 0 - 1 * (2 - 1) -1 }{2}\rfloor + 1$
   - $W_{out} = \lfloor \frac{W_{in} - 1 -1 }{2}\rfloor + 1$
   - $W_{out} = \lfloor \frac{W_{in} - 2 + 2 }{2}\rfloor$
   - $W_{out} = \lfloor \frac{W_{in}}{2}\rfloor$
   - Es decir, reduce el tamaÑo del input a la mitad, por lo que la capa de convolución reduce a las imágenes de 100x100 a 50x50, la segunda capa las reduce a 25x25, y las terceras las reduce a 12x12 (tener en cuenta la operación *floor*).
- ***Fully connected***
  - el input de esta capa será de 64x12x12. $64$ por ser la cantidad de kernels de salida de la tercer capa convolucional. $12$x$12$ se debe a la reducción que realiza el máx pooling sobre las imágenes
  - El Output será de $7$, la cantidad de clases a clasificar.

```python
def bloque_convolucion(c_in, c_out, kernel_size=3, padding='same', stride=1, pooling=2):
  return torch.nn.Sequential(
      torch.nn.Conv2d(c_in, c_out, kernel_size, padding=padding, stride=stride),
      torch.nn.BatchNorm2d(num_features=c_out),
      torch.nn.ReLU(),
      torch.nn.MaxPool2d(pooling)
  )
```

```python
class CNN(torch.nn.Module):
  def __init__(self, n_channels=1, n_outputs=7, kernel_size=3,
               padding="same", stride=1, pooling=2, dropout=0.5):
    super().__init__()
    C_OUT_1 = 16
    C_OUT_2 = 32
    C_OUT_3 = 64
    self.conv1 = bloque_convolucion(c_in = n_channels, c_out = C_OUT_1,
                                    kernel_size=kernel_size, padding=padding,
                                    stride=stride, pooling=pooling)
    self.conv1_out = None
    self.drop = torch.nn.Dropout2d(p=dropout, inplace=False)
    self.conv2 = bloque_convolucion(c_in = C_OUT_1, c_out = C_OUT_2,
                                    kernel_size=kernel_size, padding=padding,
                                    stride=stride, pooling=pooling)
    self.conv2_out = None
    self.conv3 = bloque_convolucion(c_in = C_OUT_2, c_out = C_OUT_3,
                                    kernel_size=kernel_size, padding=padding,
                                    stride=stride, pooling=pooling)
    self.conv3_out = None

    self.fully_connected = torch.nn.Linear(C_OUT_3*25*25, n_outputs)

  def forward(self, x):
    self.conv1_out = self.drop(self.conv1(x))
    self.conv2_out = self.drop(self.conv2(self.conv1_out))
    self.conv3_out = self.conv3(self.conv2_out)
    y = self.conv3_out
    y = y.flatten(start_dim=1)
    y = self.fully_connected(y)
    return y
```

### Funcion para training loop

```python
def train_model(
        model,
        optimizer,
        train_loader,
        eval_loader,
        loss_module,
        patience=5,
        patience_factor=0.01,
        num_epochs=10,
        num_outputs=7, # Cantidad de clases a predecir
        show_progress=True,
        scheduler_type="step",
        scheduler=None
):
    model.to(device)

    best_eval = 0
    pt_epoch = 0

    # Acumuladores de métricas finales
    train_loss_by_epoch=[]
    valid_loss_by_epoch=[]

    train_f1_by_epoch = []
    valid_f1_by_epoch = []

    train_accuracy_by_epoch = []
    valid_accuracy_by_epoch = []

    for epoch in tqdm(range(num_epochs), desc="Epoch Progress", disable=(not show_progress)):
        model.train()
        train_loss = 0.0
        train_f1 = 0.0
        train_correct = 0
        train_samples = 0

        train_preds = torch.tensor([])
        train_preds = train_preds.to(device)
        train_labels = torch.tensor([])
        train_labels = train_labels.to(device)

        for batch_idx, data in enumerate(train_loader):

            X, y = data

            data_inputs = X.to(device)
            data_labels = y.to(device)

            preds = model(data_inputs)
            preds = preds.to(device)
            preds = preds.squeeze(dim=1)
            preds = preds.to(device)

            loss = loss_module(preds, data_labels)

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            batch_loss = loss.item()
            train_loss += batch_loss

            if preds.ndim == 2:  # Softmax case
                preds_classes = preds.argmax(dim=1)
            else:  # Sigmoid case (binary classification)
                preds_classes = (preds > 0.5).long()

            train_correct += (preds_classes == data_labels).sum().item()
            train_samples += data_labels.size(0)

            # F1
            f1_metric = F1Score(task="multiclass", num_classes=num_outputs).to(device)
            f1_metric.update(preds_classes.detach().clone().to(device), data_labels.detach().clone().to(device))
            train_f1 += (f1_metric.compute().item())

            train_preds = torch.cat((train_preds, preds))
            train_labels = torch.cat((train_labels, data_labels))

            if scheduler_type=="step" and scheduler is not None:
                scheduler.step()

        train_loss /= len(train_loader)
        train_accuracy = train_correct / train_samples
        train_f1_epoch = train_f1  / len(train_loader)

        ########## Evaluación ##########
        model.eval()
        eval_loss = 0.0
        eval_f1 = 0.0
        eval_correct = 0
        eval_samples = 0
        all_preds = torch.tensor([])
        all_preds = all_preds.to(device)
        all_labels = torch.tensor([])
        all_labels = all_labels.to(device)

        with torch.no_grad():
            for data in eval_loader:

                X, y = data
                data_inputs = X.to(device)
                data_labels = y.to(device)

                preds = model(data_inputs)
                preds = preds.squeeze(dim=1)
                preds = preds.to(device)
                # 2
                loss = loss_module(preds, data_labels)
                eval_loss += loss.item()

                if preds.ndim == 2:
                    preds_classes = preds.argmax(dim=1)
                else:
                    preds_classes = (preds > 0.5).long()

                eval_correct += (preds_classes == data_labels).sum().item()
                eval_samples += data_labels.size(0)

                # F1
                f1_metric = F1Score(task="multiclass", num_classes=num_outputs).to(device)
                f1_metric.update(preds_classes.detach().clone().to(device), data_labels.detach().clone().to(device))
                eval_f1 += (f1_metric.compute().item())

                all_preds = torch.cat((all_preds, preds))
                all_labels = torch.cat((all_labels, data_labels))

        eval_loss /= len(eval_loader)
        eval_accuracy = eval_correct / eval_samples
        eval_f1_epoch = eval_f1  / len(eval_loader)

        ########## Return data ##########
        train_loss_by_epoch.append(train_loss)
        valid_loss_by_epoch.append(eval_loss)

        train_f1_by_epoch.append(train_f1_epoch)
        valid_f1_by_epoch.append(eval_f1_epoch)

        train_accuracy_by_epoch.append(train_accuracy)
        valid_accuracy_by_epoch.append(eval_accuracy)

        if show_progress:
          print(f"Epoch {epoch+1}/{num_epochs} - Train Loss: {train_loss:.4f} - Train Acc: {train_accuracy:.4f} | Eval Loss: {eval_loss:.4f} - Eval Acc: {eval_accuracy:.4f}")

        if scheduler_type=="epoch" and scheduler is not None:
            scheduler.step()

        ## Early stopping
        if eval_accuracy>=best_eval*(1+patience_factor):
            best_eval = eval_accuracy
            pt_epoch = 0
        else:
            pt_epoch += 1
            if pt_epoch>=patience:
                print(f"Epoch {epoch+1}/{num_epochs} - Training interrupted due to early stopping condition.")
                break
            else:
                print(f"Epoch {epoch+1}/{num_epochs} - Current epochs without validation metric improvement {pt_epoch}. {patience-pt_epoch} remaining before stopping.")

    return train_loss_by_epoch, valid_loss_by_epoch, train_f1_by_epoch, valid_f1_by_epoch, train_accuracy_by_epoch, valid_accuracy_by_epoch, all_preds, all_labels
```

### Optimización de hiperparámetros

En esta sección se define la función objetivo para optimizar los siguientes hiperparámetros:

- `kernel_size`
  - Hiperparémetro del modelo `CNN`.
  - valores estudiados: $3$, $5$, $7$, $11$, $13$, $17$.
- `learning_rate`
  - Hiperparémetro del optimizador `Adam Weighted`.
  - valores estudiados: $0.001$, $0.0005$, $0.0001$, $0.00005$, $0.00001$.
- `weight_decay`
  - Hiperparémetro del optimizador `Adam Weighted`.
  - valores estudiados: $0.01$, $0.005$, $0.001$, $0.0005$, $0.0001$.

Se utilizarán $5$ epochs, ya que esta operación suele demorar demasiado. Igualmente, la función de entrenamiento está configurada con `early stopping`, con `patience` de $2$ y `patience factor` de $0.01$.

El objetivo de esta función es `maximizar el accuracy`.

```python
def objective(trial):
    learning_rate = trial.suggest_categorical("lr", [0.001, 0.0005, 0.0001, 0.00005, 0.00001])
    weight_decay = trial.suggest_categorical("weight_decay", [0.01, 0.005, 0.001, 0.0005, 0.0001])
    kernel_size = trial.suggest_categorical("kernel_size", [3, 5, 7, 11, 13, 17])

    epochs = 5
    PATIENCE = 2
    num_outputs = 7

    model = CNN(n_channels=1, n_outputs=num_outputs, padding="same",
                stride=1, pooling=2, dropout=0.5,
                kernel_size=kernel_size # único hiperparámetro a optimizar
              )

    device = torch.device("cuda:0") if torch.cuda.is_available() else torch.device("cpu")
    model = model.to(device)

    optimizer = torch.optim.AdamW(model.parameters(),
                                  lr=learning_rate, # hiperparámetro a optimizar
                                  weight_decay=weight_decay, # hiperparámetro a optimizar
                                )
    # Loss Function con weights, por clases no balanceadas
    # loss_function = torch.nn.CrossEntropyLoss(weight=weights_train_hpt.to(device))
    loss_function = torch.nn.CrossEntropyLoss()

    # Scheduler
    scheduler_type = "step"
    steps_per_epoch = len(dataloader_train_hpt)
    scheduler = CosineAnnealingWarmRestarts(optimizer, T_0=steps_per_epoch,
                                            T_mult=2, eta_min=learning_rate/10)


    _, _, _, val_loss, val_f1, val_acc, _, _ = train_model(model=model,
                                                           optimizer=optimizer,
                                                           train_loader=dataloader_train_hpt,
                                                           eval_loader=dataloader_validation_hpt,
                                                           loss_module=loss_function,
                                                           patience=PATIENCE,
                                                           patience_factor=0.01,
                                                           num_epochs=epochs,
                                                           num_outputs=num_outputs,
                                                           show_progress=False,
                                                           scheduler_type=scheduler_type,
                                                           scheduler=scheduler)
    return np.mean(val_acc)
```

Se ejecutan $10$ trials a fin de encontrar la mejor combinación de hiperparámetros que disminuya la `pérdida` o `loss` en el dataset de `train` (utilizando una división de $70\%$ para train_hpt y $30\%$ para test_hpt.

```python
study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=10)
print("Mejores hiperparámetros:", study.best_params)
```

En general, el modelo clasifica mal a casi todas las clases, salvo a `Enojo`, la cual la clasifica correctamente, y corresponde a la segunda clase que mejor clasifica el modelo.

En el caso de `Alegría`, la cual es la clase que mejor clasifica el modelo, el error podría deberse a la imagen seleccionada, ya que al aplicar el procesamiento el modelo podría confundirla con `Enojo` (segunda mejor clase).

El caso de `Disgusto` corresponde a la clase que peor clasifica el modelo, por lo cual es esperable que la clasifique erróneamente.

Algo curioso sucede con `Seriedad` y `Sorpresa`, clases que tiene un porcentaje de reconocer correctamente de $64\%$ y $65\%$ respectivamente, las cuales clasifica como `Tristeza` ($59\%$), es decir, son clases con valores "cercanos", de ahí a que pueda confundirlas entre sí.
